# 🔍 Practical 9 & 11: Deep Learning Diagnosis & Grad-CAM Explainability (CO4 & CO5)
**Unit V:** Visualizing convolutional feature activation heatmaps to prove model decision interpretability.


```python
import sys
import torch
sys.path.append('../src')
from explainability import load_model_and_labels, explain_leaf_prediction
import matplotlib.pyplot as plt
```

### 1. Load Trained MobileNetV2 and Run Diagnostic Inference


```python
model, idx_to_class = load_model_and_labels(device='cpu')
sample_img = '../data/samples/Grape___Black_rot_sample.jpg'
diag = explain_leaf_prediction(sample_img, model, idx_to_class)
print(f'AI Diagnosis: {diag["pred_class"]} ({diag["confidence_pct"]}% Confidence)')
for item in diag['top3']:
    print(f'  • {item["class"]}: {item["confidence"]}%')
```

### 2. Display Side-by-Side Explainability Heatmaps


```python
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 5))
ax1.imshow(diag['original_img'])
ax1.set_title('Input Leaf', fontweight='bold')
ax2.imshow(diag['heatmap'], cmap='jet')
ax2.set_title('Grad-CAM Heatmap', fontweight='bold')
ax3.imshow(diag['gradcam_overlay'])
ax3.set_title(f'Attention Overlay ({diag["confidence_pct"]}%)', fontweight='bold')
for ax in [ax1, ax2, ax3]: ax.axis('off')
plt.tight_layout()
plt.show()
```

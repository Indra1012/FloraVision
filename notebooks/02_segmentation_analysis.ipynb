**Unit III:** Leaf isolation, HSV necrosis extraction, and quantitative disease severity calculation.


```python
import sys
sys.path.append('../src')
from segmentation import run_segmentation_pipeline
import matplotlib.pyplot as plt
```

### 1. Run Segmentation and Extract Quantitative Metrics


```python
sample_leaf = '../data/samples/Corn_(maize)___Common_rust__sample.jpg'
res = run_segmentation_pipeline(sample_leaf)
m = res['metrics']
print(f'Leaf Area: {m["leaf_pixels"]:,} px | Lesion Area: {m["disease_pixels"]:,} px')
print(f'Surface Area Affected: {m["severity_pct"]}% | Condition: {m["status"]}')
```

### 2. Visualize Multi-Stage Segmentation Masks


```python
fig, axes = plt.subplots(1, 4, figsize=(18, 5))
axes[0].imshow(res['original'])
axes[0].set_title('1. Original Leaf', fontweight='bold')
axes[1].imshow(res['leaf_mask'], cmap='gray')
axes[1].set_title('2. Leaf Blade Mask', fontweight='bold')
axes[2].imshow(res['disease_mask'], cmap='hot')
axes[2].set_title('3. Isolated Lesions', fontweight='bold')
axes[3].imshow(res['overlay'])
axes[3].set_title(f'4. Severity Overlay: {m["severity_pct"]}%', fontweight='bold')
for ax in axes: ax.axis('off')
plt.tight_layout()
plt.show()
```

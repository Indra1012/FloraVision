```python
import sys
from pathlib import Path
import matplotlib.pyplot as plt

# Add src to path
sys.path.append('../src')
from classical_processing import process_leaf_pipeline, load_image
```

### 1. Load Sample Leaf and Execute Classical Pipeline


```python
sample_leaf = '../data/samples/Tomato___Early_blight_sample.jpg'
results = process_leaf_pipeline(sample_leaf)
print(f'Detected ORB keypoints: {results["num_keypoints"]} points')
```

### 2. Visualize Multi-Panel Transformations (CO1 & CO2)


```python
fig, axes = plt.subplots(2, 4, figsize=(18, 9))
panels = [
    (results['original'], 'Original RGB Leaf', None),
    (results['grayscale'], 'Grayscale (CO1)', 'gray'),
    (results['hue'], 'HSV: Hue Channel (Chlorophyll)', 'hsv'),
    (results['saturation'], 'HSV: Saturation (Lesions)', 'inferno'),
    (results['clahe'], 'CLAHE Enhanced (CO2)', None),
    (results['sobel_mag'], 'Sobel Gradient Magnitude (CO2)', 'magma'),
    (results['canny'], 'Canny Edge Map (CO2)', 'gray'),
    (results['orb'], f'ORB Keypoints: {results["num_keypoints"]} pts (CO2)', None),
]
for ax, (img, title, cmap) in zip(axes.flat, panels):
    ax.imshow(img, cmap=cmap)
    ax.set_title(title, fontweight='bold')
    ax.axis('off')
plt.tight_layout()
plt.show()
```
